# Analytical SQL Query Optimization

This notebook demonstrates refactoring unoptimized analytical SQL queries using 3 core engineering techniques:
1. **Task 1: SELECT * to Explicit Columns** (reducing memory & network I/O).
2. **Task 2: Applying WHERE Filters Before JOINs** (shrinking intermediate datasets prior to join operations).
3. **Task 3: Refactoring Nested Subqueries with CTEs** (improving readability, testability, and modularity).
4. **Task 4 & 5: Benchmarking & Technical Follow-Up Q&A**.

```python
import os
import pandas as pd
from sqlalchemy import create_engine

db_path = '../analytics.db'
if not os.path.exists(db_path):
    db_path = 'analytics.db'

engine = create_engine(f'sqlite:///{db_path}')
print(f"Connected to database: {db_path}")
```

## Task 1: SELECT * vs. Explicit Column Selection

```python
def load_query(query_name):
    path = f'../queries/{query_name}.sql'
    if not os.path.exists(path):
        path = f'queries/{query_name}.sql'
    with open(path, 'r') as f:
        return f.read()

task1_sql = load_query('opt_task1_explicit_columns')
task1_df = pd.read_sql(task1_sql, engine)
print(f"Explicit Column Query Result ({task1_df.shape[1]} columns fetched):")
print(task1_df.head())
```

## Task 2: Apply Filters Before JOINs

```python
task2_sql = load_query('opt_task2_early_filtering')
task2_df = pd.read_sql(task2_sql, engine)
print(f"Early Join Filtered Result ({len(task2_df)} rows):")
print(task2_df.head())
```
