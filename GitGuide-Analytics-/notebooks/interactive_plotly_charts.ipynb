```python
import os
import pandas as pd
import plotly.graph_objects as go
from sqlalchemy import create_engine

db_path = '../analytics.db'
if not os.path.exists(db_path):
    db_path = 'analytics.db'

engine = create_engine(f'sqlite:///{db_path}')
print(f"Connected to database: {db_path}")
```

## Task 1: Building Interactive Plotly Figure with Custom Hover

```python
df_daily = pd.read_sql("SELECT * FROM daily_sales LIMIT 10", engine)
fig = go.Figure(data=go.Scatter(
    x=df_daily['order_date'],
    y=df_daily['amount'],
    mode='lines+markers',
    hovertemplate='<b>Date: %{x}</b><br>Revenue: $%{y:,.2f}<extra></extra>'
))
fig.update_layout(title='Daily Revenue Trend (Notebook View)', height=400)
fig.show()
```
