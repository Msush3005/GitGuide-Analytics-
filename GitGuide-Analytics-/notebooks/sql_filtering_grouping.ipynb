# SQL Filtering, Grouping & Aggregation Workflow

This notebook demonstrates mastering essential SQL reporting clauses:
1. **WHERE Filtering**: Pre-grouping data quality and date window checks.
2. **GROUP BY & Aggregation**: Slicing metrics across multiple dimensions (`customer_type`, `month`).
3. **HAVING Filtering**: Post-aggregation filtering on computed metric thresholds.
4. **WHERE + HAVING Combined**: Real-world query combining row quality and group thresholds.
5. **ORDER BY Ranking & Window Functions**: Sorting, `RANK() OVER ()`, and Percentage Share of Total.

```python
import os
import pandas as pd
from sqlalchemy import create_engine

db_path = '../analytics.db'
if not os.path.exists(db_path):
    db_path = 'analytics.db'

engine = create_engine(f'sqlite:///{db_path}')
print(f"Connected to database: {db_path}")
```

## Task 1: WHERE Pre-Grouping Row Filtering

```python
def load_query(query_name):
    path = f'../queries/{query_name}.sql'
    if not os.path.exists(path):
        path = f'queries/{query_name}.sql'
    with open(path, 'r') as f:
        return f.read()

where_sql = load_query('where_filtering')
where_df = pd.read_sql(where_sql, engine)
print("Task 1: WHERE Filtered Customers (Top 5):")
print(where_df.head())
```

## Task 2: Multi-Dimensional GROUP BY Aggregation

```python
groupby_sql = load_query('group_by_aggregation')
groupby_df = pd.read_sql(groupby_sql, engine)
print("Task 2: GROUP BY Customer Type & Month (Top 5):")
print(groupby_df.head())
```

## Task 3: HAVING Post-Aggregation Group Filtering

```python
having_sql = load_query('having_filtering')
having_df = pd.read_sql(having_sql, engine)
print("Task 3: HAVING Filtered Groups (Revenue > $4,000 & 5+ Tx):")
print(having_df.head())
```

## Task 4: WHERE + HAVING Combined Query

```python
combined_sql = load_query('where_having_combined')
combined_df = pd.read_sql(combined_sql, engine)
print("Task 4: WHERE + HAVING Combined Segment Summary:")
print(combined_df)
```

## Task 5: ORDER BY & Window Ranking

```python
ranking_sql = load_query('order_by_ranking')
ranking_df = pd.read_sql(ranking_sql, engine)
print("Task 5: Top Industry-Segment Rankings (ORDER BY total_revenue DESC):")
print(ranking_df.head(10))

pct_sql = load_query('percentage_share')
pct_df = pd.read_sql(pct_sql, engine)
print("\nPercentage Share of Total Revenue per Segment:")
print(pct_df)
```
