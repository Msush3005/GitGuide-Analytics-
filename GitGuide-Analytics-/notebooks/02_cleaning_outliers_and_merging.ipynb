## 2. Statistical Outlier Detection (Z-Score & IQR)
Detect numerical outliers beyond 3 standard deviations and 1.5x IQR bounds.


```python
df_outliers = pd.read_csv('../data/raw/outlier_data.csv')
print('Raw Outlier Data:')
print(df_outliers)

# Calculate IQR Bounds for Revenue
q1 = df_outliers['revenue'].quantile(0.25)
q3 = df_outliers['revenue'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
print(f'IQR Bounds for Revenue: [{lower_bound:.2f}, {upper_bound:.2f}]')

# Cap Outliers
df_outliers['revenue_capped'] = df_outliers['revenue'].clip(lower=lower_bound, upper=upper_bound)
print('After Capping:')
print(df_outliers[['customer_id', 'revenue', 'revenue_capped']])
```

## 3. Relational Table Merging & Join Diagnostics
Merge customer records with order transactions on customer_id and validate row count deltas.


```python
df_customers = pd.read_csv('../data/raw/customers_merge.csv')
df_orders = pd.read_csv('../data/raw/orders_merge.csv')
print(f'Customers: {len(df_customers)} rows, Orders: {len(df_orders)} rows')

df_merged = pd.merge(df_customers, df_orders, on='customer_id', how='left')
print(f'Merged Result: {len(df_merged)} rows')
print(df_merged.head())
```
