## 1. Setup & Environment Imports
Import required libraries for dataset intake, schema validation, and exploration.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style='darkgrid')
```

## 2. Raw Data Ingestion & Schema Audit
Load raw customer and transaction datasets explicitly logging row counts and column types.


```python
raw_customers_path = '../data/raw/customer_activity.csv'
df_raw = pd.read_csv(raw_customers_path)
print(f'Raw Records Loaded: {len(df_raw)}')
df_raw.info()
df_raw.head()
```

## 3. Missing Value Analysis & Null Distribution
Check for missing values across raw columns.


```python
null_counts = df_raw.isnull().sum()
null_pct = (df_raw.isnull().sum() / len(df_raw)) * 100
null_summary = pd.DataFrame({'Null Count': null_counts, 'Null Percentage (%)': null_pct})
print(null_summary)
```

## 4. Initial Feature Distributions
Visualize initial distributions of tenure, transactions, and spend.


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.histplot(df_raw['days_as_customer'], kde=True, ax=axes[0], color='skyblue').set(title='Tenure Distribution (Days)')
sns.histplot(df_raw['total_transactions'], kde=True, ax=axes[1], color='teal').set(title='Total Transactions Distribution')
sns.histplot(df_raw['total_spent'], kde=True, ax=axes[2], color='coral').set(title='Total Spent Distribution ($)')
plt.tight_layout()
plt.show()
```
