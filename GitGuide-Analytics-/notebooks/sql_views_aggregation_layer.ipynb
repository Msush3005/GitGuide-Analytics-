# SQL Views & Aggregation Layer Design

This notebook demonstrates building a clean data layer:
1. **Task 1: Creating SQL Views** (`vw_active_customers`, `vw_product_performance`) to establish a single source of truth.
2. **Task 2: Pre-Aggregated Summary Tables** (`agg_daily_metrics`) with `updated_at` timestamps for high-speed reporting.
3. **Task 3: Dashboard Querying & Performance Benchmarking**.
4. **Task 4 & 5: Version-Controlled `.sql` Files and Naming Conventions**.

```python
import os
import pandas as pd
from sqlalchemy import create_engine

db_path = '../analytics.db'
if not os.path.exists(db_path):
    db_path = 'analytics.db'

engine = create_engine(f'sqlite:///{db_path}')
print(f"Connected to database: {db_path}")
```

## Task 1: Querying SQL Views (`vw_` Single Source of Truth)

```python
view1_df = pd.read_sql("SELECT * FROM vw_active_customers LIMIT 10", engine)
print("View 1: vw_active_customers:")
print(view1_df.head())

view2_df = pd.read_sql("SELECT * FROM vw_product_performance LIMIT 10", engine)
print("\nView 2: vw_product_performance:")
print(view2_df.head())
```

## Task 2: Pre-Aggregated Summary Table (`agg_daily_metrics`)

```python
agg_df = pd.read_sql("SELECT * FROM agg_daily_metrics ORDER BY aggregation_date DESC LIMIT 10", engine)
print("Pre-Aggregated Daily Metrics Table:")
print(agg_df)
```
