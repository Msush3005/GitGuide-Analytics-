## 1. Setup & Environment Imports


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style='white')
```

## 2. Load Customer Churn Dataset & Inspect Schema


```python
df = pd.read_csv('../data/raw/churn_customer_data.csv')
print(f'Records Loaded: {len(df)}')
df.head()
```

## 3. Pearson vs. Spearman Correlation Analysis
Calculate Pearson (linear) and Spearman (monotonic) correlations with churn target.


```python
df_numeric = df.drop(columns=['customer_id'], errors='ignore')
pearson_corr = df_numeric.corr(method='pearson')
spearman_corr = df_numeric.corr(method='spearman')

comparison = pd.DataFrame({'pearson': pearson_corr['churn'], 'spearman': spearman_corr['churn']})
print('Churn Correlation Comparison:')
print(comparison.round(4))
```

## 4. Visualize Correlation Heatmap Matrix


```python
plt.figure(figsize=(10, 8))
sns.heatmap(pearson_corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, linewidths=0.5)
plt.title('Feature Correlation Matrix (Pearson)', fontsize=14)
plt.tight_layout()
plt.show()
```

## 5. Identify Strongly Correlated Pairs (|r| > 0.7)


```python
corr_flat = pearson_corr.unstack()
strong = corr_flat[corr_flat.abs() > 0.7].sort_values(ascending=False)
strong_pairs = strong[strong != 1.0]
print('Strongly Correlated Pairs:')
print(strong_pairs)
```

## 6. Feature Selection: Drop Redundant Collinear Features
Drop `engagement` (r=0.92 with `transactions_per_month`) to preserve model interpretability.


```python
df_features = df_numeric[['transactions_per_month', 'engagement', 'support_tickets', 'churn']].copy()
print('Before Dropping Redundant Feature:')
print(df_features.corr().round(3))

df_features_selected = df_features.drop(columns=['engagement'])
print('\nAfter Dropping Engagement:')
print(df_features_selected.corr().round(3))
```
