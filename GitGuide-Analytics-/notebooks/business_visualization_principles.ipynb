# Business Visualisation Principles

This notebook demonstrates creating 5 distinct business visualizations adhering to core visualization principles:
1. **Task 1: Creating 5 Chart Types** (Bar Chart, Line Chart, Histogram, Stacked Bar, Scatter Plot).
2. **Task 2: Complete Labelling** (Titles, X/Y axes with units, legends, human-readable numbers like `$1.2M`).
3. **Task 3: Consistent Company Palette** (`PALETTE` dictionary applied consistently).
4. **Task 4: Annotating Key Insights** (`ax.annotate()`, `ax.axhline()`).
5. **Task 5: Exporting High-Resolution 300DPI Images & Documentation**.

```python
import os
from IPython.display import Image, display

output_dir = '../output' if os.path.exists('../output') else 'output'
charts = [
    'chart1_revenue_by_product.png',
    'chart2_revenue_trend.png',
    'chart3_order_value_distribution.png',
    'chart4_revenue_composition.png',
    'chart5_marketing_vs_revenue.png'
]

for chart_name in charts:
    path = os.path.join(output_dir, chart_name)
    if os.path.exists(path):
        print(f"Displaying {chart_name}:")
        display(Image(filename=path))
    else:
        print(f"Chart not found: {path}")
```
